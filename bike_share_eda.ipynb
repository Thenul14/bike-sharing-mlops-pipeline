```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import preprocessing, model_selection
from sqlalchemy import create_engine

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("bike_share/hour.csv")
df.head()
```

```python
from src.config import database_url
eng_conn = create_engine(database_url())
```

```python
df.to_sql("bike_share", con=eng_conn, if_exists="replace", index=False)
```

```python
data = pd.read_sql("bike_share", con=eng_conn)
data.head()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data['cnt'], bins=40, kde=True)
plt.title("Distribution of Bike Rentals In The Dataset")
plt.xlabel("Total Rentals")
plt.show()
```

```python
data['cnt'].skew()
```

```python
sns.boxplot(y=data['cnt'])
plt.title('Boxplot of Bike Rentals', fontsize=16, fontweight='bold')
plt.ylabel('Total Rentals', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
data["temp_celsius"] = df["temp"] * 47 - 8
data["atemp_celsius"] = df["atemp"] * 66 - 16
data["humidity_percent"] = df["hum"] * 100
data["windspeed_actual"] = df["windspeed"] * 67
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data['temp_celsius'], kde=True)
plt.xlabel("Temperature")
plt.title("Temperature Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data['atemp_celsius'], kde=True)
plt.xlabel("Temperature")
plt.title("Temperature Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data['humidity_percent'], kde=True)
plt.xlabel("Temperature")
plt.title("Temperature Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data['windspeed_actual'], kde=True)
plt.xlabel("Temperature")
plt.title("Temperature Distribution")
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.scatterplot(x='temp_celsius', y='cnt', data=data)
plt.title("Temperature vs Bike Rentals")
plt.show()
```
